# Word Embeddings: One-Hot Encoding and Word2Vec

**Learning objective:** Compare sparse one-hot representations with dense Word2Vec vectors learned from a small example corpus.

**Environment:** Jupyter Notebook or Google Colab. Run cells from top to bottom. Install commands are included where needed.

---

## Learning outcomes

- Build one-hot word representations.
- Train a small Word2Vec model on an example corpus.
- Explain the difference between sparse and dense representations.


```python
from sklearn.preprocessing import OneHotEncoder
```

```python
words = [['cat'],['dog'],['apple'],['car']]
```

```python
encoder = OneHotEncoder(sparse_output=False)
result = encoder.fit_transform(words)
```

```python
print("words :",encoder.categories_[0])
print("Embeddings:")
print(result)
```

```python
from gensim.models import Word2Vec
```

```python
sentences = [["python","is","a","language"],
 ["java","is","a","language"],
  ["python","is","easy"],
   ["java","is","easy"],
    ["python","is","powerful"],
     ["java","is","powerful"]]
```

```python
model = Word2Vec(sentences,vector_size=10,window=2,min_count=1,workers=1)
```

```python
print(model.wv.most_similar('java',topn=3))
```

```python
print(model.wv.most_similar('python',topn=4))
```
